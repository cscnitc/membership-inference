```python
cat_cols = [c for c in df.select_dtypes(include='str').columns
            if c != 'readmitted' and 2 <= df[c].nunique(dropna=False) <= 20]

fig, axes = plt.subplots(len(cat_cols), 1, figsize=(20, len(cat_cols)))

for ax, c in zip(axes, cat_cols):
    # Fill NaN with an explicit "Missing" category instead of letting
    # groupby silently drop those rows (pandas groupby defaults to dropna=True,
    # which would hide rows with missing race/diag values from this plot).
    grouped_col = df[c].fillna('Missing')
    rates = (df['readmitted'].groupby(grouped_col)
               .apply(lambda s: (s != 'NO').mean())
               .sort_values(ascending=False))

    ax.imshow(rates.values.reshape(1, -1), aspect='auto',
              cmap='Oranges', vmin=0, vmax=1)
    ax.set_xticks(range(len(rates)),
                  [f"{k}\n{v:.0%}" for k, v in rates.items()], fontsize=15)
    ax.set_yticks([])
    ax.set_ylabel(c, rotation=20, fontsize=15)

plt.tight_layout()
plt.show()
```

The above Heatmap shows the relation between the every feature's unique set to the readmitted column.
This show how much percentage a given feature is for being readmitted according to the data.

Dark color => higher readmission rate for that category.

Light color => lower readmission rate.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, c in zip(axes.flat, ['time_in_hospital', 'num_medications', 'number_inpatient', 'number_diagnoses']):
    ax.hist(df.loc[df['readmitted']=='NO',  c], bins=30, alpha=0.5, density=True, label='NO')
    ax.hist(df.loc[df['readmitted']!='NO', c], bins=30, alpha=0.5, density=True, label='YES')
    ax.set_title(c)
    ax.legend()
plt.tight_layout()
plt.show()
```

### Implementing the above solutions to dataframe

```python
# Drop columns that are constant (no predictive value) or have
# excessive missing values / low predictive value for this task.
# encounter_id is kept for now -- it's needed one more step, to determine
# each patient's first encounter below, and gets dropped right after.
cols_to_drop = ['weight', 'medical_specialty', 'payer_code', 'examide', 'citoglipton']
df = df.drop(columns=cols_to_drop)
df.shape
```

```python
# Keep only each patient's first encounter, so the same patient can
# never appear in more than one data split later (e.g. target train vs.
# shadow-out pool) -- letting that happen would leak information across
# the split and invalidate the membership inference experiment.
#
# Sort by encounter_id within each patient as a proxy for chronological
# order (encounter_id is roughly time-ordered in this dataset), then keep
# only the first row per patient_nbr.
df = df.sort_values(by=['patient_nbr', 'encounter_id'])
df = df.drop_duplicates(subset='patient_nbr', keep='first')
df = df.reset_index(drop=True)
df.shape
```

```python
# encounter_id has served its purpose (picking the first encounter above)
# and is dropped now -- it's a row identifier, not a predictive feature.
df = df.drop(columns=['encounter_id'])
df.shape
```
